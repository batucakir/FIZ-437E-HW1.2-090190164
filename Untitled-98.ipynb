```python
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
import geopandas as gpd
from shapely.geometry import Point, box

try:
    from adjustText import adjust_text
    _USE_ADJUST_TEXT = True
except:
    _USE_ADJUST_TEXT = False

try:
    import contextily as cx
    _HAS_CTX = True
except:
    _HAS_CTX = False

# Nokta normalizasyonu (lat-lon sırası ters ise düzeltir)
def norm_point(p):
    if isinstance(p, dict):
        lat = float(p.get("lat"))
        lon = float(p.get("lon") or p.get("lng"))
        name = str(p.get("name", ""))
        grp  = (p.get("group") or "").lower() or None
    else:
        vals = list(p)
        lat, lon, name = float(vals[0]), float(vals[1]), str(vals[2])
        grp = str(vals[3]).lower() if len(vals) >= 4 else None
    if 25 <= lat <= 46.5 and 34 <= lon <= 43.8:
        lon, lat = lat, lon
    if grp not in {"a","b","c"}: grp = "other"
    return lat, lon, name, grp

def make_gdf(points):
    rows = [norm_point(p) for p in points]
    geoms = [Point(lon, lat) for lat, lon, _, _ in rows]
    names = [r[2] for r in rows]
    grps  = [r[3] for r in rows]
    return gpd.GeoDataFrame({"name":names, "group":grps}, geometry=geoms, crs="EPSG:4326")

# Estetik parametreler
COLORS = {"a":"#ff8c00", "b":"#2ca02c", "c":"#000000", "other":"#1f77b4"}  # turuncu/yeşil/siyah/mavi
EDGE   = "#101010"
```

```python
# safer, internetten yükleme
import geopandas as gpd

def load_world():
    """Natural Earth verisini internetten yükle (yerel dataset olmadan)."""
    url = "https://raw.githubusercontent.com/geopandas/geopandas/main/geopandas/datasets/naturalearth_lowres/naturalearth_lowres.shp.zip"
    world = gpd.read_file(url)
    return world

def draw_static_map(gdf_pts, outfile_png="turkey_points_pretty.png", style="positron",
                    dpi=300, label_size=10):
    # Komşular + Türkiye
    world = gpd.read_file(gpd.datasets.get_path("naturalearth_lowres"))
    neighbors = ["Turkey","Greece","Bulgaria","Georgia","Armenia","Azerbaijan","Syria","Iraq","Iran"]
    gdf_c = world[world["name"].isin(neighbors)].to_crs(3857)
    pts3857 = gdf_pts.to_crs(3857)

    # Türkiye bbox + tampon (görüntü çerçevesi)
    tur = gdf_c[gdf_c["name"]=="Turkey"]
    minx, miny, maxx, maxy = tur.to_crs(4326).total_bounds
    padx, pady = 2.5, 2.0
    xmin, ymin, xmax, ymax = (minx-padx, miny-pady, maxx+padx, maxy+pady)
    bbox_poly = gpd.GeoSeries([box(xmin, ymin, xmax, ymax)], crs=4326).to_crs(3857)
    bxmin, bymin, bxmax, bymax = bbox_poly.total_bounds

    # Şık figür
    fig = plt.figure(figsize=(12, 9), dpi=dpi)
    ax = plt.gca()

    # 1) Taban harita (varsa)
    if _HAS_CTX:
        tile = {
            "positron": cx.providers.CartoDB.Positron,
            "toner":    cx.providers.Stamen.TonerLite,
            "terrain":  cx.providers.Stamen.TerrainBackground,
            "osm":      cx.providers.OpenStreetMap.Mapnik,
        }.get(style, cx.providers.CartoDB.Positron)
        cx.add_basemap(ax, source=tile, zoom=6)  # detaylı ve temiz görünüm

    # 2) Ülkeler (taban harita yoksa daha şık göster)
    face = "#d9dde3" if not _HAS_CTX else (0,0,0,0)  # tile varsa saydam
    gdf_c.plot(ax=ax, color=face, edgecolor="#585f69", linewidth=0.8, zorder=2)
    tur.plot(ax=ax, color="#d0d4da" if not _HAS_CTX else (0,0,0,0), edgecolor="#3a4149", linewidth=1.0, zorder=3)

    # 3) Noktalar
    for grp, sub in pts3857.groupby("group"):
        sub.plot(ax=ax, markersize=36, color=COLORS.get(grp, COLORS["other"]),
                 edgecolor=EDGE, linewidth=0.5, alpha=0.98, zorder=6)

    # 4) Etiketler (halo + yön çizgisi, çakışmayı azalt)
    texts = []
    for x, y, name in zip(pts3857.geometry.x, pts3857.geometry.y, pts3857["name"]):
        t = ax.text(
            x+9000, y+9000, name, fontsize=label_size, color="#111",
            ha="left", va="bottom",
            path_effects=[pe.withStroke(linewidth=2.8, foreground="white", alpha=0.95)]
        )
        texts.append(t)
    if _USE_ADJUST_TEXT:
        adjust_text(texts, ax=ax, arrowprops=dict(arrowstyle="-", lw=0.6, color="#50565f"))

    # 5) Legend (minimal)
    legend_items = [("A", COLORS["a"]), ("B", COLORS["b"]), ("C", COLORS["c"]), ("Diğer", COLORS["other"])]
    for i, (lbl, col) in enumerate(legend_items):
        ax.scatter([], [], s=80, color=col, edgecolor=EDGE, linewidth=0.6, zorder=10,
                   label=lbl)
    leg = ax.legend(frameon=True, loc="lower left", fontsize=9)
    leg.get_frame().set_alpha(0.85)

    ax.set_xlim(bxmin, bxmax); ax.set_ylim(bymin, bymax)
    ax.set_axis_off()
    plt.tight_layout()

    plt.savefig(outfile_png, dpi=dpi, bbox_inches="tight")
    plt.savefig(outfile_png.replace(".png",".svg"), bbox_inches="tight")
    plt.show()
    print("✅ Kaydedildi:", outfile_png, "ve SVG versiyonu")
```
